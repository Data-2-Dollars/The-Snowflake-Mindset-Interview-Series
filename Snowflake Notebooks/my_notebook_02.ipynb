```
import streamlit as st
import pandas as pd
import numpy as np

# Sample data generation
np.random.seed(42)
dates = pd.date_range("2026-01-01", periods=180)
data = pd.DataFrame({
    "Date": dates,
    "Units_Sold": np.random.randint(10, 100, size=180),
    "Revenue": np.random.randint(1000, 5000, size=180)
})

# Render interactive tabs
tab1, tab2 = st.tabs(["📊 Revenue Trend", "📦 Units Sold"])

with tab1:
    st.subheader("Revenue Over Time")
    st.line_chart(data.set_index("Date")["Revenue"])

with tab2:
    st.subheader("Daily Units Sold")
    st.bar_chart(data.set_index("Date")["Units_Sold"])

# Interactive Date Slider
selected_range = st.slider(
    "Select Date Index Range", 
    min_value=0, 
    max_value=len(data)-1, 
    value=(0, 30)
)
st.write(f"Showing data for indices: {selected_range}")
```
